```python
import numpy as np
dados_amostras = np.array([
    [1.5, 55],  # [Altura, Peso]
    [1.7, 65],
    [1.6, 60],
    [1.9, 110], # Amostra com defeito (outlier)
    [1.8, 70]
])
imc = (dados_amostras[:,1]/(dados_amostras[:,0])**2)
print(imc)
dadosok = dados_amostras[ dados_amostras[:,1]<100]
print(dadosok)
media=np.mean(dadosok[:,1])
print(media)
mediana=np.median(dadosok[:,1])
print(mediana)
```
